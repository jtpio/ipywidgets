```python
import threading
from time import sleep

import ipywidgets as widgets

out = widgets.Output()

def write_to_output():
    with out:
        for i in range(10):
            print(i)
            sleep(0.5)

display(out)
```

```python
threading.Thread(target=write_to_output).start()
```
